### **BIBLIOTECA-PERSONAL**

```python
%%writefile calest.py

def media(datos):
  suma = 0
  for x in datos:
    suma = suma + x
  return suma/len(datos)

def de(datos):
  m = media(datos)
  suma_cuadrados = 0
  for x in datos:
    suma_cuadrados = suma_cuadrados + (x - m)**2
  return (suma_cuadrados/len(datos)**0.5)
```

### **BIBLIOTECA-PANDAS**

```python
import pandas as pd

df = pd.read_excel("sae_multidimensional_2024.xlsx")

val = df[df["Región"] == "Valparaíso"]["Pobreza"]
lr = df[df["Región"] == "Los Ríos"]["Pobreza"]

print("Valparaíso")
print("Media    ", round(val.mean(),1))
print("Desviación estandar   ", round(val.std(),1))

print("\nLos Ríos")
print("Media    ", round(lr.mean(),1))
print("Desviación estandar   ", round(lr.std(),1))
```
